- **Write a 100-word paragraph to describe the relationship between**
  - **`Geodatabase`**
  - **`Feature Dataset`**
  - **`Feature Classes`**
  - I can refer to **Page 9** of [assignment guide](https://miamioh.instructure.com/courses/246872/files/folder/week-02-resources?preview=38195670) or other resources online, such as
    - [Fundamentals of geodatabase](https://pro.arcgis.com/en/pro-app/latest/help/data/geodatabases/overview/fundamentals-of-the-geodatabase.htm)
    - [Feature datasets in ArcGIS Pro](https://pro.arcgis.com/en/pro-app/latest/help/data/feature-datasets/feature-datasets-in-arcgis-pro.htm)
- The geodatabase is a structured organization system tha store a collection of datasets. The dataset held in the geodatabase is a fundamental mechanism for organizing and using geospatial data. Inside each feature dataset are the feature classes, which are the individual layers representing various features to use for your map. In our practice test, the roads and lakes data were the two layers that were in the feature class. It is within the feature classes that attributes and tables of information regarding the layers are held. The relationship between all three allows for consistency and a structured way to be able to access and share your data and projects.
